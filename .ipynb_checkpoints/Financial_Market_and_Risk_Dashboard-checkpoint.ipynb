```python
!pip install yfinance
import numpy as np
import pandas as pd
import yfinance as yf #yahoo finance
import matplotlib.pyplot as plt
```

```python
actifs=["AAPL", "MSFT", "AMZN", "^GSPC", "TTE.PA", "JNJ"] 
start_date="2005-01-01"
end_date="2026-01-01"
```

```python
# Télécharger les prix de clôture ajustés
data=yf.download(actifs,start_date,end_date, auto_adjust=True)["Close"]
data.head()
```

```python
# 2. Rendements journaliers
daily_returns = data.pct_change(fill_method=None).dropna() 
```

```python
# 3. Rendements cumulés (performance nette)
cumulative_returns = (1 + daily_returns).cumprod() - 1
```

```python
# 4. Volatilité annualisée (252 jours de bourse par an)
annualized_volatility = daily_returns.std() * np.sqrt(252)
```

```python
# 5. Maximum Drawdown (MDD)
rolling_max = (1 + daily_returns).cumprod().cummax()
drawdown = (1 + daily_returns).cumprod() / rolling_max - 1
max_drawdown = drawdown.min()
```

```python
annualized_return=daily_returns.mean()*252
sharpe_ratio = annualized_return / annualized_volatility
```

```python
# 7. Matrice de corrélation
correlation_matrix = daily_returns.corr()
```

```python
#8. SIMULATION DE PORTEFEUILLE PERSONNALISÉ & RAPPORT
weights = np.array([0.2, 0.2, 0.2, 0.2,0.1,0.1]) 

portfolio_returns = daily_returns.dot(weights)
portfolio_cumulative = (1 + portfolio_returns).cumprod()
pourcentage= (portfolio_cumulative-1)*100


portfolio_cumulative.plot(figsize=(10, 5), title="Évolution de la valeur d'un portefeuille investi")
plt.ylabel("Valeur du portefeuille")
plt.show()
```

```python
report_df = pd.DataFrame({
    'Rendement Quotidien Portefeuille': portfolio_returns,
    'Valeur Cumulative': portfolio_cumulative,
    'Pourcentage de gain' : pourcentage
})
```

```python
summary_df = pd.DataFrame(
    {
        "Rendement Annualisé": annualized_return,
        "Volatilité Annualisée": annualized_volatility,
        "Ratio de Sharpe": sharpe_ratio,
        "Max Drawdown": max_drawdown,
    }
)
print("--- Résumé des Performances ---")
print(summary_df)
print("\n--- Matrice de Corrélation ---")
print(correlation_matrix)
print ("\n ---Portefeuille--- ")
print(report_df)
```
